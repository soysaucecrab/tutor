```python
import numpy as np
import matplotlib.pyplot as plt

# 원래 평면상의 점들
X = np.linspace(-3,3,11)
Y = np.linspace(-3,3,11)
pts = np.array([(x,y) for x in X for y in Y])

# g(원래 점) = projection onto x-axis
g_pts = np.array([(x,0) for (x,y) in pts])

# f(g(x, y)) = f(projection) = (x,0) → same as g_pts
fg_pts = g_pts # since f(x)=(x,0) and g(x,y)=(x,0)

plt.figure(figsize=(8,8))

# 원래 점들
plt.scatter(pts[:,0], pts[:,1], c='lightgray', label='Original points (R^2)')

# g(원래 점), projection
plt.scatter(g_pts[:,0], g_pts[:,1], c='blue', label='g(x,y)=x projection')

for p, q in zip(pts, g_pts):
    plt.arrow(p[0], p[1], q[0]-p[0], q[1]-p[1], 
              head_width=0.05, length_includes_head=True, alpha=0.3)

plt.legend()
plt.title("Visualizing g∘f = id vs f∘g ≠ id")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define x range
x = np.linspace(-3, 3, 400)

# Functions
y_f = -x**3          # f(x) = x^3
y_inv = -np.cbrt(x)  # inverse function f^{-1}(x) = cube root of x
y_neg = -x          # y = -x

# Plot
plt.figure()
plt.plot(x, y_f, label="y = -x^3")
plt.plot(x, y_inv, label="y = -∛x")
plt.plot(x, y_neg, linestyle="--", label="y = -x")

plt.axhline(0)
plt.axvline(0)
plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title("Intersections of y = x^3 and its inverse")

plt.show()
```
